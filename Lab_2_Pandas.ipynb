```python
import pandas as pd

df = pd.DataFrame({
    'Marks': [80, 90, 80, 70, 100]
})

print("Mean:", df['Marks'].mean())
print("Median:", df['Marks'].median())
print("Mode:", df['Marks'].mode()[0])
```

```python
import pandas as pd

df = pd.DataFrame({
    'Department': ['CSE', 'ECE', 'CSE', 'ECE'],
    'Marks': [80, 70, 90, 85]
})

result = df.groupby('Department')['Marks'].mean()

print(result)
```

```python
import pandas as pd

df1 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Name': ['Ali', 'Rahul', 'Priya']
})

df2 = pd.DataFrame({
    'ID': [1, 2, 3],
    'Marks': [85, 90, 95]
})

result = pd.merge(df1, df2, on='ID')

print(result)
```
