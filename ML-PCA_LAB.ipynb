```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# Load dataset directly
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/wine-quality/winequality-red.csv"
df = pd.read_csv(url, sep=";")

print("Dataset shape:", df.shape)
print(df.head())

# Select 11 input features
X = df.drop(columns=["quality"])

# Standardize the features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Apply PCA with 2 components
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

print("Original features:", X.shape[1])
print("Reduced features:", X_pca.shape[1])

print("Explained variance ratio:",
      pca.explained_variance_ratio_)

print("Total variance retained:",
      pca.explained_variance_ratio_.sum() * 100, "%")

# Plot the two principal components
plt.figure(figsize=(8, 5))
plt.scatter(
    X_pca[:, 0],
    X_pca[:, 1],
    c=df["quality"],
    cmap="viridis",
    alpha=0.7
)
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("PCA on Wine Quality Dataset")
plt.colorbar(label="Wine Quality")
plt.show()
```

## 1. Are the classes clearly separated?

Ans : No. The three wine quality classes generally overlap in the PCA plot. This indicates that the first two principal components do not completely distinguish wines by quality.

## 2. What information might be lost when reducing 13 features to 2?

## 3. Does PCA use the class labels to create the components?

Ans : No. PCA is an unsupervised dimensionality reduction technique. It uses the numerical input features to identify directions of maximum variance. The wine quality labels are used only to color the plot, not to calculate the principal components

## 2. What is the difference between feature selection and feature extraction?
ANS:
- Feature selection: Selects a subset of the original features without changing them. For example, choosing only alcohol, pH, and density from the 11 input features.
- Feature extraction: Transforms the original features into a new set of features. For example, converting 11 original features into 2 principal components using PCA.

## 3. Why is PCA considered a feature extraction technique?

Ans : PCA is considered a feature extraction technique because it creates new features called principal components by calculating linear combinations of the original features. These components capture as much of the dataset's variance as possible, in descending order.
